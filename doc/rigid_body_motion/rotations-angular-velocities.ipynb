# Rotations and Angular Velocities


## Rotation Matrices

Assume that a length scale for physical space has been chosen, and that both the fixed frame ${s}$ and body frame ${b}$ have been chosen as shown. All reference frames are right-handed – the unit axes $\mathbf{\{\hat{x}, \hat{y}, \hat{z}\}}$ always satisfy $\mathbf{\hat{x}}$ cross $\mathbf{\hat{y}}$ equals to $\mathbf{{\hat{z}}}$. Denote the unit axes of the fixed frame by $\mathbf{\{\hat{x}_{s}, \hat{y}_{s}, \hat{z}_{s}\}}$ and the unit axes of the body frame by $\mathbf{\{\hat{x}_{b}, \hat{y}_{b}, \hat{z}_{b}\}}$. Let $\mathbf{p}$ denote the vector from the fixed-frame origin to the body-frame origin. In terms of the fixed-frame coordinates, $\mathbf{p} $can be expressed as

\begin{equation*}
p = p_{1} \cdot \mathbf{\hat{x}_{s}} + p_{2} \cdot \mathbf{\hat{y}_{s}} + p_{3} \cdot \mathbf{\hat{z}_{s}}
\end{equation*}

The axes of the body frame can also be expressed as

\begin{equation*}
\mathbf{\hat{x}_{b}} = r_{11} \cdot \mathbf{\hat{x}_{s}} + r_{21} \cdot \mathbf{\hat{y}_{s}} + r_{31} \cdot \mathbf{\hat{z}_{s}}
\end{equation*}

\begin{equation*}
\mathbf{\hat{y}_{b}} = r_{12} \cdot \mathbf{\hat{x}_{s}} + r_{22} \cdot \mathbf{\hat{y}_{s}} + r_{32} \cdot \mathbf{\hat{z}_{s}}
\end{equation*}

\begin{equation*}
\mathbf{\hat{z}_{b}} = r_{13} \cdot \mathbf{\hat{x}_{s}} + r_{23} \cdot \mathbf{\hat{y}_{s}} + r_{33} \cdot \mathbf{\hat{z}_{s}}
\end{equation*}

Defining $p \in \mathbb{R}^3$, and $R \in \mathbb{R}^{3\times 3}$ as

\begin{equation*}
   p = \begin{bmatrix}
       p_{1} &
       p_{2} &
       p_{3}
       \end{bmatrix}
\end{equation*}

\begin{equation*}
        R = \begin{bmatrix}
            \mathbf{\hat{x}_{b}} &
            \mathbf{\hat{y}_{b}} &
            \mathbf{\hat{z}_{b}}
            \end{bmatrix}
            = \begin{bmatrix}
            r_{11} &
            r_{12} &
            r_{13} \\
            r_{21} &
            r_{22} &
            r_{23} \\
            r_{31} &
            r_{32} &
            r_{33} \\
            \end{bmatrix}
\end{equation*}

the 12 parameters given by $(R, p)$ then provide a description of the position and orientation of the rigid body relative to the fixed frame. Since the orientation of a rigid body has **three degrees of freedom**, only three of the nine entries in $R$ can be chosen independently.

We argued earlier that, of the nine entries in the rotation matrix R, only three can be chosen independently. We begin by expressing **a set of six explicit constraints** on the entries of $R$. These six constraints can be expressed more compactly as a single set of constraints on the matrix $R$, 

\begin{equation*}
    R^{T}R = I
\end{equation*}

where $R^{T}$ denotes the transpose of $R$, and $I$ denotes the identity matrix.

The six equality constraints represented imply that $det{R} = \pm 1$; imposing the additional constraint $det{R} = 1$ means that only **right-handed frames are allowed**. This additional constraint does not change the number of independent continuous variables needed to parametrize $R$.

**Definition. The special orthogonal group $SO(3)$**, also known as the group of rotation matrices, is the set of all $3\times 3$ real matrices $R$ that satisfy (i) $R^{T}R = I$ and (ii) $det{R} = 1$. The set of $2\times 2$ rotation matrices is a subgroup of $SO(3)$ and is denoted $SO(2)$.

## Angular Velocities


Referring to the following figure, suppose that a frame with unit axes $\{ \hat{x}, \hat{y}, \hat{z} \}$ is attached to a rotating body. The angular velocity $\omega$ is defined as follows:

\begin{equation*}
    \omega = \hat{\omega} \cdot \dot{\theta}
\end{equation*}

where $\hat{\omega}$ is the unit axis of the rotation, and $\dot{\theta}$ represents the ratio $\Delta \theta / \Delta t$ when the limit as $\Delta t$ approaches zero.

It should be evident that the linear velocity 

\begin{equation*}
    \dot{\hat{x}} = \omega \times \hat{x}
\end{equation*}

\begin{equation*}
    \dot{\hat{y}} = \omega \times \hat{y}
\end{equation*}

\begin{equation*}
    \dot{\hat{z}} = \omega \times \hat{z}
\end{equation*}

Let us start with fixed-frame ${s}$ coordinates. Let $R(t)$ be the rotation matrix describing the orientation of the body frame w.r.t the fixed frame at time $t$:

\begin{equation*}
        R(t) = \begin{bmatrix}
            r_{1}(t) &
            r_{2}(t) &
            r_{3}(t)
            \end{bmatrix}
\end{equation*}

At a specific time $t$, let $\omega_{s} \in \mathbb{R}^3$ be the angular velocity expressed in fixed-frame coordinate. Those three equations can be rearranged into:

\begin{equation*}
        \dot{R}(t) = \begin{bmatrix}
            \omega_{s} \times r_{1}(t) &
            \omega_{s} \times r_{2}(t) &
            \omega_{s} \times r_{3}(t)
            \end{bmatrix}
            = \omega_{s} \times R(t)
\end{equation*}
    
To eliminate the cross product here, we introduce some new notation, rewriting $\omega_{s} \times R(t)$ as $[\omega_{s}] R(t)$, where $[\omega_{s}]$ is a $3 \times 3$ **skew-symmetric** matrix representation of $\omega_{s} \in \mathbb{R}^3$.

**Definition. Skew-Symmetric Matrix Representation** 

Given a vector $x = [x_{1} \quad x_{2} \quad x_{3}]^{T} \in \mathbb{R}^{3}$, define
\begin{equation*}
    [x] = \begin{bmatrix}
        0 & -x_{3} & x_{2} \\
        x_{3} & 0 & -x_{1} \\
        -x_{2} & x_{1} & 0 \\
    \end{bmatrix}
\end{equation*}

The matrix $[x]$ is a $3 \times 3$ skew-symmetric matrix representation of $x$; that is,

\begin{equation*}
    [x] = -[x]^{T}
\end{equation*}

The set of all $3 \times 3$ real skew-symmetric matrices is called $so(3)$. 

The set of skew-symmetric matrices $so(3)$ is called the **Lie algebra** of the **Lie group** $SO(3)$.

- The Lie algebra $so(3)$ represents the tangent space at the identity matrix $I$ of the Lie group $SO(3)$. This means $so(3)$ describes all possible infinitesimal rotations around the identity matrix.
- For a small rotation, the change in a rotation matrix $R$ can be expressed as $[\omega_{s}] R(t)$, where $\dot{R}(t)$ is in $so(3)$ when $R$ is the identity matrix $I$.

Given any $\omega \in \mathbb{R}^{3} and R \in SO(3)$, the following always holds:

\begin{equation*}
R[\omega]R^{T} = [R\omega]
\end{equation*}

With the skew-symmetric notation, we can rewrite angular velocity equation as:

\begin{equation*}
[\omega_{s}]R(t) = \dot{R}(t)
\end{equation*}

We can post-multiply both sides of the equation by $R^{-1}$ to get

\begin{equation*}
[\omega_{s}] = \dot{R}(t)R(t)^{-1}
\end{equation*}

Now let $^{s}\omega_{b}$ express the body-frame $\omega$ with respect to fixed-frame coordinate. Since we have $^{s}\omega_{b} = ^{s}R_{b} ^{s}\omega_{s}$, where $^{s}\omega_{b}$ represents the axis of rotation in body-frame w.r.t fixed frame, $^{s}R_{b}$ is the body-frame orientation w.r.t fixed-frame and $^{s}\omega_{s}$ is the axis of rotation in fixed-frame w.r.t fixed-frame(no need).

## Exponential Coordinate Representation of Rotation

The exponential coordinates parametrize a rotation matrix in terms of a rotation axis (unit vector $\hat{\omega}$) and an angle of rotation $\theta$ about that axis; the vector $\hat{\omega} \cdot \theta \in \mathbb{R}^{3}$ then serves as the three-parameter **exponential coordinate representation** of the rotation. Writing $\hat{\omega}$ and $\theta$ individually is the **axis-angle** representation of a rotation.

Let us begin with the simple scalar linear differential equation, 
\begin{equation*}
\dot{x}(t) = ax(t)
\end{equation*}

where $x(t) \in \mathbb{R}$, $a \in \mathbb{R}$ is constant, and the initial condtion $x(0) = x_{0}$ is given. Then it has solution:

\begin{equation*}
x(t) = e^{at}x_{0}
\end{equation*}

It is also useful to remember the series expansion of the exponential function:

\begin{equation*}
e^{at} = 1 + at + \frac{(at)^{2}}{2!} + \frac{(at)^{3}}{3!} + \ldots
\end{equation*}

Now consider the vector linear differential equation

\begin{equation*}
\dot{x}(t) = Ax(t)
\end{equation*}

where $x(t) \in \mathbb{R}^{n}$, $A \in \mathbb{R}^{n \times n}$ is constant, and the initial condtion $x(0) = x_{0}$ is given. From the above scalar result one can conjecture a solution of the form

\begin{equation*}
x(t) = e^{At}x_{0}
\end{equation*}

where the **matrix exponential** $e^{At}$ now needs to be defined in a meaningful way. Again mimicking the scalar case, we define the matrix exponential to be

\begin{equation*}
e^{At} = 1 + at + \frac{(At)^{2}}{2!} + \frac{(At)^{3}}{3!} + \ldots
\end{equation*}

The first question to be addressed is under what conditions this series converges, so that the matrix exponential is well-defined. It can be shown that if $A$ is constant and finite then this series is always guaranteed to converge to a finite limit; the proof can be found in most texts on ordinary linear differential equations and is not covered here.

The second question is to simply prove $\dot{x}(t) = Ax(t)$ by taking the time derivative of $x(t) = e^{At}x_{0}$

\begin{equation*}
\begin{align}
\dot{x}(t) &= \left( \frac{d}{dt}e^{At} \right) x_{0} \\
           &= \frac{d}{dt} \left( I + At + \frac{A^{2}t^{2}}{2!} + \frac{A^{3}t^{3}}{3!} + \ldots \right) x_{0} \\
           &= \left(  A + A^{2}t + \frac{A^{3}t^{2}}{2!} + \ldots \right) x_{0} \\
           &= Ae^{At}x_{0} \\
           &= Ax(t)
\end{align}
\end{equation*}

which proves that $x(t) = e^{At}x_{0}$ is indeed a solution. That this is a unique solution follows from the basic existence and uniqueness result for linear ordinary differential equations, which we invoke here without proof.

While $AB \neq BA$ for arbitrary square matrices A and B, it is always true that

\begin{equation*}
Ae^{At} = e^{At}A
\end{equation*}

for any square $A$ and scalar $t$. You can verify this directly using the series expansion for the matrix exponential. Therefore, we could also get

\begin{equation*}
\dot{x}(t) = e^{At}Ax_{0}
\end{equation*}

While the matrix exponential $e^{At}$ is defined as an infinite series, closed-form expressions are often available. For example, if $A$ can be expressed as $A = PDP^{-1}$ for some $D\in \mathbb{R}^{n\times n}$ and invertible $p\in \mathbb{R}^{n\times n}$ then

\begin{equation*}
\begin{align}
e^{At} &= 1 + At + \frac{(At)^{2}}{2!} + \ldots \\
       &= I + (PDP^{-1})t + (PDP^{-1})(PDP^{-1})\frac{t^{2}}{2!} + \ldots \\
       &= P(I + Dt + \frac{(Dt)^{2}}{2!} + \ldots )P^{-1} \\
       &= Pe^{Dt}P^{-1}
\end{align}
\end{equation*}

if moreover $D$ is diagonal, i.e., $D = diag \left[ d_{1}, d_{2}, \ldots , d_{n} \right]$, then its matrix exponential is particularly simple to evaluate: $e^{Dt} = diag \left[ e^{d_{1}t}, e^{d_{2}t}, \ldots e^{d_{n}t} \right]$
